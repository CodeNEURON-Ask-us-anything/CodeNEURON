```python
from transformers import pipeline
nli = pipeline("text-classification", model="facebook/bart-large-mnli")

print("Model Loaded Successfully")
```

```python

premise = "Python uses the Timsort algorithm for sorting lists."
hypothesis = "Python uses Timsort for sorting."

result = nli(f"{premise} </s></s> {hypothesis}")

print("Entailment Test Output:")
print(result)
```

```python

premise = "Canberra is the capital city of Australia."
hypothesis = "The capital of Australia is Sydney."

result = nli(f"{premise} </s></s> {hypothesis}")

print("Contradiction Test Output:")
print(result)
```
